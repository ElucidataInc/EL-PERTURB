```python
import pandas as pd
from pathlib import Path
import re

for cline in ['k562', 'rpe1', 'jurkat', 'hepg2', 'K562GW']:

    all_data = []
    for f in Path(f"gears_eval_results/{cline}").glob(f"*_summary.csv"):
        # Extract metadata from filename
        # Adjust regex to match your pattern
        match = re.search(r'(\w+)_TF_(\d+)_UF_\d+_rs_(\d+)_(\w+)', f.stem)
        if match:
            variant = 'gears'
            cellline, tf, seed, split = match.groups()
            df = pd.read_csv(f)
            df = df.loc[:, ["Group", "N", "Mean_MSE", "Mean_WMSE", "Mean_Pearson", "Mean_Pearson_Pert", "Mean_Pearson_TopK_DEGs", "Mean_Rank_L1", "Mean_Rank_Cosine", "Matrix_Dist", "Global_Score", "Specificity_Score"]]
            df['variant'] = variant
            df['train_frac'] = int(tf)
            df['seed'] = int(seed)
            df['split'] = split
            all_data.append(df)
    
    merged = pd.concat(all_data, ignore_index=True)
    merged.to_csv(f"gears_{cline}_all_metrics_merged.csv", index=False)
```
